# Track 2 — Anomaly Triage Pipeline: Telemetry Exploration
Profile `baseline_telemetry.csv` and inspect seeded faults in `fault_injected_stream.csv` before implementing `src/detector.py` and `src/router.py`.

```python
import csv
from pathlib import Path
from collections import Counter

DATA_DIR = Path("../data/synthetic")
with (DATA_DIR / "baseline_telemetry.csv").open() as f:
    baseline = list(csv.DictReader(f))
with (DATA_DIR / "fault_injected_stream.csv").open() as f:
    stream = list(csv.DictReader(f))

print(f"Baseline rows: {len(baseline)} | Stream rows: {len(stream)}")
print("Stream states:", Counter(r["state"] for r in stream))
print("Max baseline latency (ms):", max(int(r["latency_ms"]) for r in baseline))
print("Max stream latency (ms):", max(int(r["latency_ms"]) for r in stream))
```
